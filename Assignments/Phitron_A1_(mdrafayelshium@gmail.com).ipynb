Question-1

```python
lst=input('Enter numbers: ').split()

output=[]

for i in lst:
  intgr=int(i)
  if intgr%4==0:
    if intgr%8!=0:
      output.append(intgr)

print(output)

```

Question-2

```python
def ratio(x,y):

    try:
        c=(x+y)/(x*y)

    except ZeroDivisionError:

          print("Error:Division not possible")

    else:
      return c

lst=input("Enter two numbers: ").split()

r=ratio(int(lst[0]),int(lst[1]))
if r is not None:
    print(r)
```

```python
with open('/content/sample_data/data.txt') as file:

  words=file.read().split()
  longest_word=""

for word in words:
    if len(word)>len(longest_word):
        longest_word=word

print(longest_word)

```

Question-4

```python
lst=input("Enter the sentence: ").split()

words_with_vowel=[]

for i in lst:
  if i[0].lower() in ['a','e','i','o','u']:
    words_with_vowel.append(i)

print(words_with_vowel)
```

```python
lst=input("Enter word: ").lower()

flags=True

for i in range(len(lst)//2):

  if lst[i]!=lst[len(lst)-1-i]:
    flags=False
    break

if flags==True:
  print('Palindrome')
else:
  print('Not Palindrome')
```
